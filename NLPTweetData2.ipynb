```python
from transformers import pipeline
```

```python
chosen_model = "distilbert-base-uncased-finetuned-sst-2-english"
distil_bert = pipeline(task="sentiment-analysis", model=chosen_model, device = 0, batch_size = 32)
```

```python
print(df["Clean_Text"].head())
print(df["Clean_Text"].apply(type).value_counts())
```

```python
df["Clean_Text"] = df["Clean_Text"].astype("str")
```

```python
print(df["Clean_Text"].str.len().value_counts())
```

```python
import torch
torch.cuda.is_available()
```

```python
df[df["Clean_Text"].str.len() < 10]
```

```python
df1 = df[(df["Clean_Text"].str.len() > 10) & (df["Clean_Text"] != "nan")]
```

```python
display(df1.shape)
display(df.shape)
display(df1["Clean_Text"].str.len().value_counts())
```

```python
results = distil_bert(df1["Clean_Text"].tolist())
```

```python
results
```

```python
df1["Sentiment"] = [result["label"] for result in results]
df1["Score"] = [result["score"] for result in results]
```
